# 例6-8 rc参数设置示例1

```python
import numpy as np
import matplotlib.pyplot as plt
plt.rcParams['font.family'] = ['SimHei']
plt.rcParams['axes.unicode_minus'] = False

def f(t):
    return np.cos(2 * np.pi * t)

x1 = np.arange(0.0,4.0,0.5)
x2 = np.arange(0.0,4.0,0.01)
plt.figure(1)
plt.subplot(2,2,1)
plt.plot(x1,f(x1),'bo',x2,f(x2),'k')
plt.title('子图1')
plt.subplot(2,2,2)
plt.plot(np.cos(2 * np.pi * x2),'r--')
plt.title('子图2')
plt.show()
```

# 例6-9 rc参数设置示例2

```python
fig = plt.figure()
ax = fig.add_subplot(1,1,1)
ax.plot(np.random.randn(30).cumsum(), color = 'k', linestyle = 'dashed', marker = 'o', label = 'one')
ax.plot(np.random.randn(30).cumsum(), color = 'k', linestyle = 'dashed', marker = '+', label = 'two')
ax.plot(np.random.randn(30).cumsum(), color = 'k', linestyle = 'dashed', marker = 'v', label = 'three')
ax.legend(loc = 'best')
```

# 例6-10 用set_xticks设置X轴刻度

```python
fig = plt.figure()
ax = fig.add_subplot(1,1,1)
ax.plot(np.random.randn(30).cumsum(), color = 'k', linestyle = 'dashed', marker = 'o', label = 'one')
ax.plot(np.random.randn(30).cumsum(), color = 'k', linestyle = 'dashed', marker = '+', label = 'two')
ax.plot(np.random.randn(30).cumsum(), color = 'k', linestyle = 'dashed', marker = 'v', label = 'three')
ax.legend(loc = 'best')
ax.set_xticks([0,5,10,15,20,25,30,35])
```

# 例6-12 使用fill_between()填充区域

```python
import numpy as np
import matplotlib.pyplot as plt
x = np.linspace(0,1,500)
y = np.sin(3 * np.pi * x) * np.exp(-4 * x)
fig,ax = plt.subplots()
plt.plot(x,y)
plt.fill_between(x, 0, y, facecolor = 'green', alpha = 0.3)
plt.show()
```

# 例6-15 直接使用fill进行绘图的填充

```python
import numpy as np
import matplotlib.pyplot as plt
x = np.linspace(0,1,500)
y = np.sin(3 * np.pi * x) * np.exp(-4 * x)
fig,ax = plt.subplots()
ax.fill(x,y)
plt.show()
```
