```python
import jieba
from wordcloud import WordCloud
import matplotlib.pyplot as plt

def get_wordCloud():
    # 直接读取你的春.txt，修复编码
    file = open("春.txt", encoding="utf-8")
    mylist = file.read()
    file.close()
    
    # 分词
    word_list = jieba.cut(mylist)
    new_text = ' '.join(word_list)
    
    # 生成词云（不需要任何图片、停用词文件）
    wordcloud = WordCloud(
        background_color="white",
        font_path='C:/Windows/Fonts/msyh.ttc',  #  Windows 自带字体
        width=800,
        height=600
    ).generate(new_text)
    
    # 显示
    plt.imshow(wordcloud)
    plt.axis("off")
    plt.show()

get_wordCloud()
```

```python
import jieba
from wordcloud import WordCloud
import PIL.Image as image
import numpy as np
import matplotlib.pyplot as plt

def get_wordCloud():
    file = open("春.txt", encoding="utf-8")
    mylist = file.read()
    file.close()
    
    word_list = jieba.cut(mylist)
    new_text = ' '.join(word_list)
    pic_path = 'hudie.png'
    img_mask = np.array(image.open(pic_path))
    
    wordcloud = WordCloud(
        background_color="white",
        font_path='C:\\Windows\\Fonts\\msyh.ttc',
        mask=img_mask
    ).generate(new_text)
    
    plt.imshow(wordcloud)
    plt.axis("off")
    plt.show()

get_wordCloud()
```
