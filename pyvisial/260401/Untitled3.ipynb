```python
from pyecharts import options as opts
from pyecharts.charts import Pie
import pandas as pd
provinces = ['北京', '上海', '黑龙江', '吉林', '辽宁', '内蒙古', '新疆', '西藏', '青海',
             '四川', '云南', '陕西', '重庆', '贵州', '广西', '海南', '澳门', '湖南', '江西',
             '福建', '安徽', '浙江', '江苏', '宁夏', '山西', '河北', '天津']

num = [1, 1, 1, 17, 9, 22, 23, 42, 35, 7, 20, 21, 16, 24,16, 21, 37, 12, 13, 14,
       13, 7, 22, 8, 16, 13, 13] 
color_series = ['#FAE927', '#E9E416', '#C9DA36', '#9ECB3C', '#6DBC49',
                '#37B44E', '#3DBA78', '#14ADCF', '#209AC9', '#1E91CA',
                '#2C6A00', '#2B55A1', '#2D3D8E', '#44388E', '#6A368B',
                '#7D3990', '#A63F98', '#C31C88', '#D52178', '#D5225B',
                '#D02C2A', '#D44C2D', '#F57A34', '#FABF2F', '#D9D210',
                '#CF7B25', '#CF7B25']

# 创建DataFrame
df = pd.DataFrame({'provinces': provinces, 'num': num})
df.sort_values(by='num', ascending=False, inplace=True)

v = df['provinces'].values.tolist()
d = df['num'].values.tolist()

# 绘制玫瑰饼图
c = Pie(init_opts=opts.InitOpts(width='1250px', height='750px'))
c.set_colors(color_series)

c.add("", [list(z) for z in zip(v, d)],
      radius=["30%", "100%"],
      center=["50%", "50%"],
      rosetype="area"
      )

# 全局配置
c.set_global_opts(
    title_opts=opts.TitleOpts(
        title='多省区\n确诊病例连续多日',
        subtitle='零新增',
        title_textstyle_opts=opts.TextStyleOpts(font_size=25, color='#0085c3'),
        subtitle_textstyle_opts=opts.TextStyleOpts(font_size=50, color='#003399'),
        pos_right='center', pos_left='center', pos_top='42%'
    ),
    legend_opts=opts.LegendOpts(is_show=False),
    toolbox_opts=opts.ToolboxOpts()
)

# 系列配置
c.set_series_opts(label_opts=opts.LabelOpts(
    is_show=True, position="inside", font_size=12,
    formatter="{b}:{c}天", font_style="italic",
    font_weight="bold", font_family="SimHei"
))

# 生成HTML文件
c.render("shixun1.html")
```
