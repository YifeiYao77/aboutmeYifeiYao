```python
import pandas as pd
import numpy as np
np.random.seed(971)
v1=np.random.randint(10,size=8)
v2=np.random.randint(10,size=8)
years=np.arange(2010,2018)
groups=['A','B','D','C','A','A','B','C']
df=pd.DataFrame({'group':groups,'year':years,'v1':v1,'v2':v2})
display(df)
print('group的取值为：',df.group.unique())
print('group的取值个数为：',df.group.nunique())
print('v1的取值以及个数为：\n',df.v1.value_counts())
```
