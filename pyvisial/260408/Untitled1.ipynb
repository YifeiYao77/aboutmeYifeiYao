```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import fmin, fminbound
def f(x):
    return x ** 2 + 10 * np.sin(x) + 1
x = np.linspace(-10, 10, num=500)
min1 = fmin(f, 3)
min2 = fmin(f, 0)
min_global = fminbound(f, -10, 10)
print("极小值点 min1 (3附近):", min1)
print("极小值点 min2 (0附近):", min2)
print("全局最小值点 min_global:", min_global)
plt.plot(x, f(x))
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
x = np.linspace(-5, 5, 20)
y = 3.5 * x + 2.1
y_noise = y + np.random.randn(20) * 2
coeff = np.polyfit(x, y_noise, 1)
plt.plot(x, y_noise, 'x', label='Noisy Data')
plt.plot(x, coeff[0] * x + coeff[1], label='Fitted Line')
plt.plot(x, y, '--', label='True Line')
plt.legend()
plt.show()
print("拟合系数 (斜率, 截距):", coeff)
```

```python
from scipy import optimize
import numpy as np
def myfunc(x, a, b, w, t):
    return a * np.exp(-b * x) * np.sin(w * x + t)
x = np.linspace(0, 2 * np.pi)
par = [3,2,1.25, np.pi/4]
y = myfunc(x, *par)
y_noise = y + 0.8 * np.random.randn(len(y))
def errf(p, x, y):
    return y - myfunc(x, *p)
c,rv=optimize.leastsq(errf,[1,1,1,1],args=(x,y_noise))
plt.plot(x,y_noise,'x',x,y,x,myfunc(x,*c),':')
plt.legend(['data','actual','leastsq'])
plt.show()
```

```python
import numpy as np
from scipy import sparse
sparse.coo_matrix((2,3))
A = sparse.coo_matrix([[1, 2, 0], [0, 1, 0], [2, 0, 0]])
print( A)
print(type(A))
A_csr = A.tocsr()
print(type(A_csr))
v = np.array([1, 3, -3])
result = A_csr * v
print(result)
```
