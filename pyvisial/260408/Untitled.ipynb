```python
w = pd.date_range('2019/1/1', '2019/6/1', freq='ME')
y = pd.Series(np.arange(5), index=w)
ps = y.to_period()
print(y)
print(ps)
```

```python
w = pd.date_range(start='2018/12/28', periods=100, freq='D')
y = pd.Series(np.arange(100), index=w)
print(y.head(8))
ps=y.resample('ME').mean()
print(ps)
```

```python
wdate = pd.date_range(start='2019/5/1', periods=10, freq='D')
w = pd.Series(np.arange(10), index=wdate)
print(w)
y = w.resample('3D', closed='right', label='right').sum()
print(y)
```

```python
ydata = [datetime(2019, 6, 1), datetime(2019, 6, 6)]
y = pd.Series([1, 6], index=ydata)
print(y)
yf= y.resample('D').ffill()
print(yf)
```
